```python
import math
import matplotlib.pyplot as plt
from utils import PLOT_COLORS, TEXT_SIZE

SAVE_PLOTS = True


def shapley_weight(coalition_size: int, number_of_players: int) -> float:
    return (math.factorial(coalition_size)
            * math.factorial(number_of_players - coalition_size - 1)
            / math.factorial(number_of_players))


def banzhaf_weight(number_of_players: int) -> float:
    return 1 / 2 ** (number_of_players - 1)


number_of_players = 8
x_axis = list(range(number_of_players))
shapley_y_axis = [shapley_weight(coalition_size, number_of_players) for coalition_size in x_axis]
banzhaf_y_axis = [banzhaf_weight(number_of_players) for _ in x_axis]

fig, ax = plt.subplots(figsize=(6, 4))
ax.set_xlabel("Coalition size", fontsize=TEXT_SIZE, labelpad=15)
ax.set_ylabel("Weight", fontsize=TEXT_SIZE, labelpad=15)
ax.plot(x_axis, shapley_y_axis, color=PLOT_COLORS.BLACK, linestyle="dashed", marker="o", linewidth=2.5, markersize=7, label="Shapley")
ax.plot(x_axis, banzhaf_y_axis, color=PLOT_COLORS.GRAY, linestyle="solid", marker="o", linewidth=2.5, markersize=7, label="Banzhaf")
ax.set_xticks(x_axis)
ax.set_xticklabels([str(size) for size in x_axis], fontsize=TEXT_SIZE)
ax.tick_params(axis="y", labelsize=TEXT_SIZE)
ax.legend(fontsize=TEXT_SIZE)
plt.tight_layout()
plt.show()
if SAVE_PLOTS:
    fig.savefig("shapley_banzhaf_weight_curve_8_players.pdf", bbox_inches="tight")
```

```python
import math
import matplotlib.pyplot as plt
from utils import PLOT_COLORS, TEXT_SIZE

SAVE_PLOTS = True

def cardinality_shapley_weight(coalition_size: int, number_of_players:int) -> float:
      return 1 / math.comb(number_of_players, coalition_size)


def cardinality_banzhaf_weight(coalition_size: int, number_of_players:int) -> float:
      return (cardinality_shapley_weight(coalition_size, number_of_players) *
              (number_of_players * math.comb(number_of_players - 1, coalition_size) / 2 ** (number_of_players - 1)))

number_of_players = 8
x_axis = list(range(number_of_players + 1))
shapley_y_axis = [cardinality_shapley_weight(coalition_size, number_of_players) for coalition_size in x_axis]
banzhaf_y_axis = [cardinality_banzhaf_weight(coalition_size, number_of_players) for coalition_size in x_axis]

fig, ax = plt.subplots(figsize=(6, 4))
ax.set_xlabel("Cardinality", fontsize=TEXT_SIZE, labelpad=15)
ax.set_ylabel("Weight", fontsize=TEXT_SIZE, labelpad=15)
ax.plot(x_axis, shapley_y_axis, color=PLOT_COLORS.BLACK, linestyle="dashed", marker="o", linewidth=2.5, markersize=7, label="Card. Shapley")
ax.plot(x_axis, banzhaf_y_axis, color=PLOT_COLORS.GRAY, linestyle="solid", marker="o", linewidth=2.5, markersize=7, label="Card. Banzhaf")
ax.set_xticks(x_axis)
ax.set_xticklabels([str(size) for size in x_axis], fontsize=TEXT_SIZE)
ax.tick_params(axis="y", labelsize=TEXT_SIZE)
ax.legend(fontsize=TEXT_SIZE)
plt.tight_layout()
plt.show()
if SAVE_PLOTS:
    fig.savefig("shapley_banzhaf_cardinality-weight_curve_8_players.pdf", bbox_inches="tight")
```
